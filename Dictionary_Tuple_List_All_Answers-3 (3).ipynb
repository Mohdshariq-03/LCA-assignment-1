```python
# 1. Dictionary operations
my_dict = {
    "name": "ABC",
    "panel": "B",
    "rollno": 34,
    "branch": "CSE",
    "semester": 1
}

print("Original dictionary:", my_dict)

my_dict["age"] = 18
print("After adding age:", my_dict)

del my_dict["semester"]
print("After deleting semester:", my_dict)

print("Student information:")
for key, value in my_dict.items():
    print(key, ":", value)
```

```python
# 2. Convert lists to dictionary
list1 = ["name", "panel", "rollno"]
list2 = ["ABC", "B", 34]

my_dict = dict(zip(list1, list2))
print(my_dict)
```

```python
# 3. Sort the elements of dictionary using sorted()
my_dict = {
    "name": "ABC",
    "panel": "B",
    "rollno": 34,
    "marks": [65, 87, 67, 94]
}

print(sorted(my_dict))
```

```python
# 4. Convert a dictionary to two lists
my_dict = {
    "name": "ABC",
    "panel": "B",
    "rollno": 34,
    "branch": "CSE"
}

keys = list(my_dict.keys())
values = list(my_dict.values())

print("List of keys:", keys)
print("List of values:", values)
```

```python
# 5. Find the mean of values of dictionary
mydict = {
    "marks1": 23,
    "marks2": 123,
    "marks3": 43,
    "marks4": 13,
    "marks5": 39
}

mean = sum(mydict.values()) / len(mydict)
print("Mean of all values:", mean)
```

```python
# 6. Dictionary operations
my_dict = {
    "name": ["Yash", "Neel", "Dev"],
    "rollno": [11, 12, 13],
    "marks": [78, 56, 98]
}

# a) Display name as Dev
print("a)", my_dict["name"][2])

# b) Display 12 roll no
print("b)", my_dict["rollno"][1])

# c) Display greatest marks
print("c)", max(my_dict["marks"]))
```

```python
# 7. Calculate frequency of letters in a string
string = input("Enter a string: ")
frequency = {}

for char in string:
    if char != " ":
        frequency[char] = frequency.get(char, 0) + 1

print(frequency)
```

```python
# 8. Quotient and remainder as a tuple
numerator = int(input("Enter numerator: "))
denominator = int(input("Enter denominator: "))

quotient = numerator // denominator
remainder = numerator % denominator

result = (quotient, remainder)
print(result)
```

```python
# 9. Find number of days between two dates
from datetime import date

d1, m1, y1 = map(int, input("Enter first date (d m y): ").split())
d2, m2, y2 = map(int, input("Enter second date (d m y): ").split())

date1 = date(y1, m1, d1)
date2 = date(y2, m2, d2)

days = abs((date2 - date1).days)
print("Number of days between the two dates:", days)
```
